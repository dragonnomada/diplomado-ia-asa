```python
import numpy
import pandas

print(f"Numpy: {numpy.__version__}")
print(f"Pandas: {pandas.__version__}")
```

```python
demandas = numpy.array([8100, 8200, 8500, 8900, 9100, 9200, 9500, 9600, 9800, 9900,
                        10000, 10100, 10200, 10500, 10800, 11000, 11200, 11500, 11800, 12000])
serie_demanda = pandas.Series(demandas)
print(serie_demanda.describe())
```

```python
print(serie_demanda.plot.density())
```

```python
print(serie_demanda.plot.box(vert=False))
```

```python
print("¿Cuál es la demanda media/promedio esperada?")
print(serie_demanda.mean())
print("¿Cuánto se desvía linealmente en promedio una demanda de la media?")
print(serie_demanda.std())
print("¿Hay sesgo en los datos o su distribución se ve centrada en la media como una campana?")
sesgo = serie_demanda.skew()
if abs(sesgo) < 0.5:
    res_sesgo = f"No hay sesgo significativo (Valor de sesgo = {sesgo:.3f}). La distribución está centrada."
elif sesgo > 0:
    res_sesgo = f"Sí, hay un sesgo positivo hacia la derecha (Valor de sesgo = {sesgo:.3f})."
else:
    res_sesgo = f"Sí, hay un sesgo negativo hacia la izquierda (Valor de sesgo = {sesgo:.3f})."
print(res_sesgo)
print("¿Cuántos picos muestra la distribución?")
frecuencias, bordes = numpy.histogram(demandas, bins='auto')
picos = 0
# Un pico es un valor que es mayor que su vecino de la izquierda y de la derecha
for i in range(1, len(frecuencias) - 1):
    if frecuencias[i] > frecuencias[i-1] and frecuencias[i] > frecuencias[i+1]:
        picos += 1
# Si los datos son muy planos y no se detecta pico intermedio, asumimos 1 pico principal (unimodal)
if picos == 0 and len(frecuencias) > 0:
    picos = 1
print(picos)
print("¿Hasta cuánta demanda tiene el 25% de las demandas?")
print(serie_demanda.quantile(0.25))
print("¿Hasta cuánta demanda tiene el 50% de las demandas?")
print(serie_demanda.quantile(0.50))
print("¿Hasta cuánta demanda tiene el 75% de las demandas?")
print(serie_demanda.quantile(0.75))
```
