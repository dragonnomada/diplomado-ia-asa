```python
# 2x + 5y -z = 9
# 3x -y + z = 14
# -x + 2y + 5z = 36

# (2 5 -1) (x)   (9)
# (3 -1 1) (y) = (14)
# (-1 2 5) (z)   (36)

# A v = b

# v = A^-1 b

A = numpy.array([
    [2, 5, -1],
    [3, -1, 1],
    [-1, 2, 5],
])

b = numpy.array([9, 14, 36])

v = numpy.linalg.inv(A).dot(b)

v # x = 3, y = 2, z = 7
```

```python
v2 = numpy.linalg.solve(A, b)

v2
```

```python
numpy.linalg.eig(A)
```

```python
edades = numpy.array([24, 35, 30, 32, 45, 12, 13, 32, 34, 54, 23, 28, 19, 22])

edades
```

```python
edades2 = edades.copy()

numpy.random.shuffle(edades2)

edades2
```

```python
m = int(edades.size/2)
muestra1 = edades2[:m]
muestra2 = edades2[m:]

muestra1, muestra2
```

```python
print("            Muestra 1   | Muestra 2")
print(f"Promedio:     {muestra1.mean():.2f}     |   {muestra2.mean():.2f}")
print(f"Desviación:   {muestra1.std():.2f}     |   {muestra2.std():.2f}")
print(f"Mínimo:       {muestra1.min():.2f}     |   {muestra2.min():.2f}")
print(f"Q1 (25%):     {numpy.quantile(muestra1, 0.25):.2f}     |   {numpy.quantile(muestra2, 0.25):.2f}")
print(f"Q2 (50%):     {numpy.quantile(muestra1, 0.50):.2f}     |   {numpy.quantile(muestra2, 0.50):.2f}")
print(f"Q3 (75%):     {numpy.quantile(muestra1, 0.75):.2f}     |   {numpy.quantile(muestra2, 0.75):.2f}")
print(f"Máximo:       {muestra1.max():.2f}     |   {muestra2.mean():.2f}")
```

```python
precios = pandas.Series([34, 45.6, 35.8, 23.2, 49.45, 41.39, 23.15, 27.2], name="Precio")

precios
```

```python
precios.plot.line(marker="o", color="gray", markerfacecolor="red")
```

```python
# Gráfica de caja
# - Muestra el mínimo y máximo como aletas
# - La caja se compone del Cuartil 1 al Cuartil 3, mostrando la mediana adentro

# Se visualiza directamente el rango (el espacio donde viven los datos dentro del eje)
# Se visualizan los cortes poblaciones (Q1, Q2, Q3)
# Se visualiza el intervalo donde vive el 50% de la población (entre Q1 y Q3)
# Se visualiza si hay sesgo en los datos: Si la mediana está más cargada a Q3 significa que los valores quieres ser altos
# y por lo tanto hay sesgo a la derecha (hacia lo alto)
# Y si la mediana se carga hacia Q1 significa que los valores quieren ser bajos y hay un sesgo a la izquierda
# (los valores se pegan hacia la izquierda)


precios.plot.box(vert=False)
```

```python
precios.describe()
```
