```python
# 2x + 5y -z = 9
# 3x -y + z = 14
# -x + 2y + 5z = 36

# (2 5 -1) (x)   (9)
# (3 -1 1) (y) = (14)
# (-1 2 5) (z)   (36)

# A v = b

# v = A^-1 b

A = numpy.array([
    [2, 5, -1],
    [3, -1, 1],
    [-1, 2, 5],
])

b = numpy.array([9, 14, 36])

v = numpy.linalg.inv(A).dot(b)

v # x = 3, y = 2, z = 7
```

## 1. Adquisición de datos

```python
archivo = open("./conjuntos/datos_demanda.txt", "r")

datos = archivo.read() # str - con todos los datos

archivo.close()
```

```python
print(datos)
```

```python
lineas = datos.split("\n")

linea_fechas = lineas[0]

linea_fechas
```

```python
import re # regular-expressions (expresiones regulares)

import datetime

patron = r"(\d{2})\/(\d{2})\/(\d{4})"

fechas = []

for match in re.finditer(patron, linea_fechas):
    if not match is None:
        #print(match)
        dia = int(match.group(1))
        mes = int(match.group(2))
        año = int(match.group(3))
        fecha = (dia, mes, año)
        fechas.append(fecha)

fecha_anterior = fechas[-1]
fecha_corregida = datetime.datetime(fecha_anterior[2], fecha_anterior[1], fecha_anterior[0]) \
    + datetime.timedelta(1) # Agregamos 1 día a la fecha anterior

fechas[-1] = (fecha_corregida.day, fecha_corregida.month, fecha_corregida.year)

fechas
```

```python
print("\n".join(lineas[2:])) # Líneas de la tercera fila (2do índice) hacía adelante
```

```python
def limpiar_coma(parte):
    return parte.replace(",", "")

registros = []

# Recorre cada línea a partir de la tercera
for linea in lineas[2:]:
    # Partimos la línea de texto en una lista de textos separados por tabulador
    partes = linea.split("\t")
    # Si no hay suficientes elementos, (se espera hora, suministros, litros, ...)
    if len(partes) <= 1:
        # Ignora el demás cómputo y solicita el siguiente elemento de la secuencia
        continue
    #print(partes)
    # La primera parte es un texto, que representa un entero con la hora
    hora = int(partes[0])
    #print(partes[1::2])
    # Para cada parte a partir de la segunda y de dos en dos, extraemos los sumistros
    suministros = [int(parte) for parte in partes[1::2]]
    #print(partes[2::2])
    # Para cada parte a partir de la tercera y de dos en dos, estraemos los litros
    litros = [int(float(limpiar_coma(parte))) for parte in partes[2::2]]

    #print(hora, suministros, litros)

    # Armado del frame (Estructurar el registro)
    # Para cada fecha, suministro y litro de las 5 fechas por cada línea
    for fecha, suministro, litro in zip(fechas, suministros, litros):
        registro = {}
        registro["fecha"] = f"{fecha[2]:02d}-{fecha[1]:02d}-{fecha[0]:02d}"
        registro["hora"] = hora
        registro["suministro"] = suministro
        registro["litro"] = litro
        #print(registro)
        # Agregamos el registro
        registros.append(registro)

demandas = pandas.DataFrame(registros)

demandas
```

```python
demandas["litro_promedio"] = demandas["litro"] / demandas["suministro"]

demandas
```
