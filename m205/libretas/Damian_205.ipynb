```python
from scipy import stats

import numpy as np
import pandas as pd

from matplotlib import pyplot
import seaborn
```

```python
# Caso de estudio: Clasificación de las flores Iris
# 1. Variedad - Discreta
# 2. Ancho Sépalo - Continua
# 3. Largo Dépalo - Continua
# 4. Ancho Pétalo - Continua
# 5. Largo Pétalo - Continua

# Creencias de simulación de datos
# SW es normal (mu_sw, sigma_sw) 
# SL es normal (mu_sl, sigma_sl)
# PW es normal (mu_pw, sigma_pw)
# PL es normal (mu_pl, sigma_pl)
# Variedad se infiere por la combinación de factores de similitud a las medias

# Creando un conjunto de datos aleatorios basado en el análisis anterior de la flor iris

# Sepal Length

#       Setosa      Versicolor  Virginica
# mean	5.006000	5.936000	6.588000
# std	0.352490	0.516171	0.635880

mu_sl = [5.006, 5.936, 6.588]
sigma_sl = [.352, .516, .636]

sl1 = np.random.normal(mu_sl[0], sigma_sl[0], 200)
sl2 = np.random.normal(mu_sl[1], sigma_sl[1], 200)
sl3 = np.random.normal(mu_sl[2], sigma_sl[2], 200)

sl = np.hstack([sl1, sl2, sl3]) # 600 registros de SL (combinando 3 creencias)

sl = pd.Series(sl, name="SL").round(1)

sl
```

```python
# Sepal Width

#       S           Ver         Vir
#mean	3.428000	2.770000	2.974000
#std	0.379064	0.313798	0.322497

mu_sw = [3.428, 2.770, 2.974]
sigma_sw = [.379, .314, .322]

sw1 = np.random.normal(mu_sw[0], sigma_sw[0], 200)
sw2 = np.random.normal(mu_sw[1], sigma_sw[1], 200)
sw3 = np.random.normal(mu_sw[2], sigma_sw[2], 200)

sw = np.hstack([sw1, sw2, sw3]) # 600 registros de SW (combinando 3 creencias)

sw = pd.Series(sw, name="SW").round(1)

sw
```

```python
# Petal Length

#       S           Ver         Vir
# mean	1.462000	4.260000	5.552000
# std	0.173664	0.469911	0.551895

mu_pl = [1.462, 4.260, 5.552]
sigma_pl = [.173, .470, .552]

pl1 = np.random.normal(mu_pl[0], sigma_pl[0], 200)
pl2 = np.random.normal(mu_pl[1], sigma_pl[1], 200)
pl3 = np.random.normal(mu_pl[2], sigma_pl[2], 200)

pl = np.hstack([pl1, pl2, pl3]) # 600 registros de PL (combinando 3 creencias)

pl = pd.Series(pl, name="PL").round(1)

pl
```

```python
# Petal Width

#       S           Ver         Vir
#mean	0.246000	1.326000	2.026000
#std	0.105386	0.197753	0.274650

mu_pw = [0.246, 1.326, 2.026]
sigma_pw = [.105, .198, .275]

pw1 = np.random.normal(mu_pw[0], sigma_pw[0], 200)
pw2 = np.random.normal(mu_pw[1], sigma_pw[1], 200)
pw3 = np.random.normal(mu_pw[2], sigma_pw[2], 200)

pw = np.hstack([pw1, pw2, pw3]) # 600 registros de PW (combinando 3 creencias)

pw = pd.Series(pw, name="PW").round(1)

pw
```

```python
varieties = []

for SL, SW, PL, PW in zip(sl, sw, pl, pw):
    I = pd.DataFrame({
        "Setosa": [
            stats.norm.pdf(SL, loc=mu_sl[0], scale=sigma_sl[0]) / stats.norm.pdf(mu_sl[0], loc=mu_sl[0], scale=sigma_sl[0]),
            stats.norm.pdf(SW, loc=mu_sw[0], scale=sigma_sw[0]) / stats.norm.pdf(mu_sw[0], loc=mu_sw[0], scale=sigma_sw[0]),
            stats.norm.pdf(PL, loc=mu_pl[0], scale=sigma_pl[0]) / stats.norm.pdf(mu_pl[0], loc=mu_pl[0], scale=sigma_pl[0]),
            stats.norm.pdf(PW, loc=mu_pw[0], scale=sigma_pw[0]) / stats.norm.pdf(mu_pw[0], loc=mu_pw[0], scale=sigma_pw[0])
        ],
        "Versicolor": [
            stats.norm.pdf(SL, loc=mu_sl[1], scale=sigma_sl[1]) / stats.norm.pdf(mu_sl[1], loc=mu_sl[1], scale=sigma_sl[1]),
            stats.norm.pdf(SW, loc=mu_sw[1], scale=sigma_sw[1]) / stats.norm.pdf(mu_sw[1], loc=mu_sw[1], scale=sigma_sw[1]),
            stats.norm.pdf(PL, loc=mu_pl[1], scale=sigma_pl[1]) / stats.norm.pdf(mu_pl[1], loc=mu_pl[1], scale=sigma_pl[1]),
            stats.norm.pdf(PW, loc=mu_pw[1], scale=sigma_pw[1]) / stats.norm.pdf(mu_pw[1], loc=mu_pw[1], scale=sigma_pw[1])
        ],
        "Virginica": [
            stats.norm.pdf(SL, loc=mu_sl[2], scale=sigma_sl[2]) / stats.norm.pdf(mu_sl[2], loc=mu_sl[2], scale=sigma_sl[2]),
            stats.norm.pdf(SW, loc=mu_sw[2], scale=sigma_sw[2]) / stats.norm.pdf(mu_sw[2], loc=mu_sw[2], scale=sigma_sw[2]),
            stats.norm.pdf(PL, loc=mu_pl[2], scale=sigma_pl[2]) / stats.norm.pdf(mu_pl[2], loc=mu_pl[2], scale=sigma_pl[2]),
            stats.norm.pdf(PW, loc=mu_pw[2], scale=sigma_pw[2]) / stats.norm.pdf(mu_pw[2], loc=mu_pw[2], scale=sigma_pw[2])
        ]
    })

    '''
    print(f"Setosa:     {I.sum(axis=0)["Setosa"]:.2f}")
    print(f"Versicolor: {I.sum(axis=0)["Versicolor"]:.2f}")
    print(f"Virginica:  {I.sum(axis=0)["Virginica"]:.2f}")
    break
    '''

    s1 = I.sum(axis=0)["Setosa"]
    s2 = I.sum(axis=0)["Versicolor"]
    s3 = I.sum(axis=0)["Virginica"]

    p1 = s1 / (s1 + s2 + s3)
    p2 = s2 / (s1 + s2 + s3)
    p3 = s3 / (s1 + s2 + s3)

    #print(f"P:  {p1:.2f} {p2:.2f} {p3:.2f}")

    variety = np.random.choice(["Setosa", "Versicolor", "Virginica"], p=[p1, p2, p3])

    #print(variety)

    varieties.append(variety)

    #print("-" * 60)
   
v = pd.Series(varieties, name="V")

v
```

```python
iris_simulado = pd.DataFrame({
    "sepal.length": sl,
    "sepal.width": sw,
    "petal.length": pl,
    "petal.width": pw,
    "variety": v,
}).sort_values("variety").reset_index().drop("index", axis=1)

iris_simulado
```

```python
iris_real = pd.read_csv("../conjuntos/iris.csv")

iris_real
```

```python
mu_real = iris_real.groupby("variety").mean()
sigma_real = iris_real.groupby("variety").std()

mu_real
```

```python
mu_simulada = iris_simulado.groupby("variety").mean()
sigma_simulada = iris_simulado.groupby("variety").std()

mu_simulada
```

```python
def proba_variety(values, index_variety, mu, sigma):
    return [
        stats.norm.pdf(values[0], loc=mu.iloc[index_variety, 0], scale=sigma.iloc[index_variety, 0]) /
        stats.norm.pdf(mu.iloc[index_variety, 0], loc=mu.iloc[index_variety, 0], scale=sigma.iloc[index_variety, 0]),
        stats.norm.pdf(values[1], loc=mu.iloc[index_variety, 1], scale=sigma.iloc[index_variety, 1]) /
        stats.norm.pdf(mu.iloc[index_variety, 1], loc=mu.iloc[index_variety, 1], scale=sigma.iloc[index_variety, 1]),
        stats.norm.pdf(values[2], loc=mu.iloc[index_variety, 2], scale=sigma.iloc[index_variety, 2]) /
        stats.norm.pdf(mu.iloc[index_variety, 2], loc=mu.iloc[index_variety, 2], scale=sigma.iloc[index_variety, 2]),
        stats.norm.pdf(values[3], loc=mu.iloc[index_variety, 3], scale=sigma.iloc[index_variety, 3]) /
        stats.norm.pdf(mu.iloc[index_variety, 3], loc=mu.iloc[index_variety, 3], scale=sigma.iloc[index_variety, 3]),
    ]
```

```python
suma_error = 0
n = 0

for i, SL, SW, PL, PW, V in iris_real.itertuples():
    #print(i, SL, SW, PL, PW, V)

    I_real = pd.DataFrame({
        "Setosa": proba_variety([SL, SW, PL, PW], 0, mu_real, sigma_real),
        "Versicolor": proba_variety([SL, SW, PL, PW], 1, mu_real, sigma_real),
        "Virginica": proba_variety([SL, SW, PL, PW], 2, mu_real, sigma_real),
    })
    I_simulada = pd.DataFrame({
        "Setosa": proba_variety([SL, SW, PL, PW], 0, mu_simulada, sigma_simulada),
        "Versicolor": proba_variety([SL, SW, PL, PW], 1, mu_simulada, sigma_simulada),
        "Virginica": proba_variety([SL, SW, PL, PW], 2, mu_simulada, sigma_simulada),
    })

    #print(I_real.round(2))
    #print(I_simulada.round(2))
    #print(I_real.sum(axis=0))
    #print(I_simulada.sum(axis=0))

    error = I_real.sum(axis=0) - I_simulada.sum(axis=0)

    #print(error)

    l = error.dot(error) # Error cuadrático

    suma_error += l
    n += 1

MSE = suma_error / n

print(f"MSE: {MSE:.2f}")
```
