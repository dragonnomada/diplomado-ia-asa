## Optimización de las creencias

```python
def proba_variety(values, index_variety, mu, sigma):
    return [
        stats.norm.pdf(values[0], loc=mu.iloc[index_variety, 0], scale=sigma.iloc[index_variety, 0]) /
        stats.norm.pdf(mu.iloc[index_variety, 0], loc=mu.iloc[index_variety, 0], scale=sigma.iloc[index_variety, 0]),
        stats.norm.pdf(values[1], loc=mu.iloc[index_variety, 1], scale=sigma.iloc[index_variety, 1]) /
        stats.norm.pdf(mu.iloc[index_variety, 1], loc=mu.iloc[index_variety, 1], scale=sigma.iloc[index_variety, 1]),
        stats.norm.pdf(values[2], loc=mu.iloc[index_variety, 2], scale=sigma.iloc[index_variety, 2]) /
        stats.norm.pdf(mu.iloc[index_variety, 2], loc=mu.iloc[index_variety, 2], scale=sigma.iloc[index_variety, 2]),
        stats.norm.pdf(values[3], loc=mu.iloc[index_variety, 3], scale=sigma.iloc[index_variety, 3]) /
        stats.norm.pdf(mu.iloc[index_variety, 3], loc=mu.iloc[index_variety, 3], scale=sigma.iloc[index_variety, 3]),
    ]
```

```python
def vector_creencia(mu, sigma, n):
    return pandas.Series(numpy.hstack([
        numpy.random.normal(mu[0], sigma[0], n),
        numpy.random.normal(mu[1], sigma[1], n),
        numpy.random.normal(mu[2], sigma[2], n)
    ]))
```

```python
def vector_inferencia(sl, sw, pl, pw, mu, sigma):
    varieties = []
    for SL, SW, PL, PW in zip(sl, sw, pl, pw):
        I = pandas.DataFrame({
            "Setosa": proba_variety([SL, SW, PL, PW], 0, mu, sigma),
            "Versicolor": proba_variety([SL, SW, PL, PW], 1, mu, sigma),
            "Virginica": proba_variety([SL, SW, PL, PW], 2, mu, sigma),
        })
        
        s1 = I.sum(axis=0)["Setosa"]
        s2 = I.sum(axis=0)["Versicolor"]
        s3 = I.sum(axis=0)["Virginica"]
    
        p1 = s1 / (s1 + s2 + s3)
        p2 = s2 / (s1 + s2 + s3)
        p3 = s3 / (s1 + s2 + s3)

        variety = None
    
        if p1 > p2:
            if p1 > p3:
                variety = "Setosa"
            else:
                variety = "Virginica"
        else:
            if p2 > p3:
                variety = "Versicolor"
            else:
                variety = "Virginica"

        varieties.append(variety)

    return pandas.Series(varieties)
```

```python
def loss(iris_real, iris_simulado):
    mu_real = iris_real.groupby("variety").mean()
    sigma_real = iris_real.groupby("variety").std()
    
    mu_simulada = iris_simulado.groupby("variety").mean()
    sigma_simulada = iris_simulado.groupby("variety").std()
    
    suma_error = 0
    n = 0
    
    for i, SL, SW, PL, PW, V in iris.itertuples():
        I_real = pandas.DataFrame({
            "Setosa": proba_variety([SL, SW, PL, PW], 0, mu_real, sigma_real),
            "Vericolor": proba_variety([SL, SW, PL, PW], 1, mu_real, sigma_real),
            "Virginica": proba_variety([SL, SW, PL, PW], 2, mu_real, sigma_real),
        })
        I_simulada = pandas.DataFrame({
            "Setosa": proba_variety([SL, SW, PL, PW], 0, mu_simulada, sigma_simulada),
            "Vericolor": proba_variety([SL, SW, PL, PW], 1, mu_simulada, sigma_simulada),
            "Virginica": proba_variety([SL, SW, PL, PW], 2, mu_simulada, sigma_simulada),
        })
        error = I_real.sum(axis=0) - I_simulada.sum(axis=0)
        l = error.dot(error)
        suma_error += l
        n += 1
    
    MSE = suma_error / n
    
    return MSE
```

```python
# Modelo Análitico Final

# 1. Creencias
# 2. Generar los datos sintéticos/simulados
# 3. Comparar los datos sintéticos/simulados contra los reales (error)
# 4. Optimizar las creencias (cambiar de creencias)
# 5. Reportar las creencias finales

# Híper-parámetro (valores de simulación)
n = 333

# Parámetros de optimización (creencias iniciales)
mu_sl =    [5,    5.9,  6.6]
sigma_sl = [0.35, 0.51, 0.63]
mu_sw =    [3.4,  2.8,  3]
sigma_sw = [0.38, 0.31, 0.32]
mu_pl =    [1.46, 4.26, 5.55]
sigma_pl = [0.17, 0.47, 0.55]
mu_pw =    [0.25, 1.32, 2.03]
sigma_pw = [0.10, 0.2,  0.28]

# Optimización por épocas
for epoca in range(100):

    sl_0 = vector_creencia(mu_sl, sigma_sl, n)
    sw_0 = vector_creencia(mu_sw, sigma_sw, n)
    pl_0 = vector_creencia(mu_pl, sigma_pl, n)
    pw_0 = vector_creencia(mu_pw, sigma_pw, n)

    mu_simulada_0 = pandas.DataFrame([mu_sl, mu_sw, mu_pl, mu_pw]).T
    sigma_simulada_0 = pandas.DataFrame([sigma_sl, sigma_sw, sigma_pl, sigma_pw]).T

    iris_simulado_0 = pandas.DataFrame({
        "sepal.length": sl_0,
        "sepal.width": sw_0,
        "petal.length": pl_0,
        "petal.width": pw_0,
        "variety": vector_inferencia(sl_0, sw_0, pl_0, pw_0, mu_simulada_0, sigma_simulada_0)
    })

    mu_sl_a = numpy.array(mu_sl) + numpy.random.normal(0, 0.01)
    mu_sw_a = numpy.array(mu_sw) + numpy.random.normal(0, 0.01)
    mu_pl_a = numpy.array(mu_pl) + numpy.random.normal(0, 0.01)
    mu_pw_a = numpy.array(mu_pw) + numpy.random.normal(0, 0.01)
    
    sigma_sl_a = numpy.array(sigma_sl) + numpy.random.normal(0, 0.001)
    sigma_sw_a = numpy.array(sigma_sw) + numpy.random.normal(0, 0.001)
    sigma_pl_a = numpy.array(sigma_pl) + numpy.random.normal(0, 0.001)
    sigma_pw_a = numpy.array(sigma_pw) + numpy.random.normal(0, 0.001)

    #print(f"SL Mu: {mu_sl} {mu_sl_a}")
    
    sl_a = vector_creencia(mu_sl_a, sigma_sl_a, n)
    sw_a = vector_creencia(mu_sw_a, sigma_sw_a, n)
    pl_a = vector_creencia(mu_pl_a, sigma_pl_a, n)
    pw_a = vector_creencia(mu_pw_a, sigma_pw_a, n)

    mu_simulada_a = pandas.DataFrame([mu_sl_a, mu_sw_a, mu_pl_a, mu_pw_a]).T
    sigma_simulada_a = pandas.DataFrame([sigma_sl_a, sigma_sw_a, sigma_pl_a, sigma_pw_a]).T

    iris_simulado_a = pandas.DataFrame({
        "sepal.length": sl_a,
        "sepal.width": sw_a,
        "petal.length": pl_a,
        "petal.width": pw_a,
        "variety": vector_inferencia(sl_a, sw_a, pl_a, pw_a, mu_simulada_a, sigma_simulada_a)
    })
    
    mse_0 = loss(iris_real, iris_simulado_0)
    mse_a = loss(iris_real, iris_simulado_a)

    print(epoca, mse_0, mse_a)
    
    if mse_a < mse_0:
        mu_sl = mu_sl_a
        sigma_sl = sigma_sl_a
        mu_sw = mu_sw_a
        sigma_sw = sigma_sw_a
        mu_pl = mu_pl_a
        sigma_pl = sigma_pl_a
        mu_pw = mu_pw_a
        sigma_pw = sigma_pw_a
```
