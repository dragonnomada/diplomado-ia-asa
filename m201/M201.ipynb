```python
# Función que determina si un punto está dentro de un circulo

# Circulo: radio, centro (ox, oy)
# Punto: px, py

def enCirculo(px, py, ox, oy, radio):
    dx = px - ox
    dy = py - oy

    d = (dx ** 2 + dy ** 2) ** 0.5

    return d <= radio
```

```python
import random

def crearPersona(nombres, edadMinima, edadMaxima):
    nombre = random.choice(nombres)
    edad = random.randint(edadMinima, edadMaxima + 1)
    return {
        "nombre": nombre,
        "edad": edad
    }
```

## 1. Ciencia de datos

> Información, datos, conjuntos de datos, estadística y el análisis de datos

**Información**

Datos no capturados (o no registrados o no medidos), que están cambiando y mantienen un efecto sobre fenómenos en la naturaleza y la realidad.

Ejemplo: Una fogata es un fenómeno donde la madera hace combustión y el fuego emerge en formas casi impredecibles que generan calor y luz.

**Datos**

Valores medidos, registrados y etiquetados de información, que permiten reconstruir un estado de información capturada en un instánte específico.

Ejemplo: Temperatura, luminocidad, radio, y centro de la fogata.

**Conjunto de datos**

Es un almacén estructurado de datos, donde podemos recuperar las características y los valores sobre cada muestra/registro/observación dispuesta.

Una observación se puede entender como un registro que captura diferentes datos relacionados en una misma entidad (muestra puntual). Cada dato dentro de la observación forma una característica.

Ejemplo: Fogatas registradas en un bosque durante el verano

**Estadísca**

Es un valor que resume muchos datos en una información puntual y útil sobre la naturaleza distribuida de todos los datos sobre la misma característica.

* **Conteo/Recuento** - El número de datos de la característica $n$
* **Suma** - El valor que suman todos datos (numéricos) $S_x = \sum_{i=1}^{n} x_i$
* **Media** - El valor promedio ideal/aritmético/ponderado de los datos (numéricos) $\overline{x} = \frac{1}{n} \sum_{i=1}^{n} x_i$
* **Varianza** - El valor cuadrático promedio sin un grado de libertad en el que se aleja cada dato de la media $VAR(x) = \frac{1}{n - 1} \sum_{i=1}^{n} (x_i - \overline{x})^2$
* **Desviación** - El valor lineal promedio en el que se aleja cada dato de la media y es resultado de la raíz cuadrada de la varianza $DESV(x) = \sqrt{VAR(x)}$
* **Mínimo** - El valor mínimo en los datos $x_{min}$
* **Máximo** - El valor máximo en los datos $x_{max}$
* **Cuartil 1** - El primer cuartil o valor umbral del 25% de la población ordenada $Q_1$
* **Cuartil 2 / Mediana** - El segundo cuartil o valor umbral del 50% de la población ordenada $Q_2$
* **Cuartil 3** - El tercer cuartil o valor umbral del 75% de la población ordenada $Q_3$
* **Rango** - La diferencia del máximo y el mínimo $L = x_{max} - x_{min}$
* **Rango Intercuartílico** - La diferencia del cuartil 3 y el cuartil 1 $IQR = Q_3 - Q_1$

**Análisis de datos**

Entender un fenómeno/evento/contexto a partir de datos y ensamblar un modelo que permita reconstruir cierta parte de la información y la realidad del fenómeno, contexto o evento de la forma más aproximada posible.

Ejemplo: Analizar el impacto ambiental de las fogatas registradas satalitalmente a lo largo de un año dentro del bosque estudiado.

* Cuestionar los datos disponibles
* Planteamiento de hipótesis
* Ensamblar un modelo que explique el comportamiento estudiado

Se cuentan con 5 registros a lo largo del año, los cuales tienen información sobre la fecha, temperatura, luminosidad, radio y centro de las fogatas.

Supuestos:

* Una fogata tiene un tiempo de vida promedio de 6 horas
* Una fogata produce un impacto ambiental que depende positivamente de la temperatura, luminosidad, radio, estación climática y ubicación
* Modelo que explica el impacto ambiental del aire
* Modelo que explica el impacto ambiental de la zona (árboles, suelo, fauna)

$$
Impacto_{aire} = \frac{Temperatura}{luminosidad} \cdot radio
$$

## 2. Estadística Descriptiva e Inferencial

**Estadística descriptiva**: Reunir la información actual para formular modelos explicativos probabilísticos

* Determinar la probabilidad de tener una temperatura media
* Determinar la probabilidad de que una temperatura supere a la media
* Determinar la probabilidad de que una temperatura sea menor a un límite (ej. > 95°C)
* Determinar la probabilidad de que una temperatura sea alta, dado que su radio es pequeño

**Estadística inferencial**: Propone modelos de predicción/inferencia a partir de los datos, sus estadísticas y análisis de tendencias y patrones derivados

* Un modelo lineal que indique cómo aumenta/disminuye la temperatura cuando cambia el radio
* Un modelo no-lineal que indique cómo aumenta/disminuye la temperatura cuando cambia el radio
* Determinar la confianza/credibilidad de un modelo
* Comparar dos modelos para determinar cuánta información logran capturar o superar
* Medir el error en pronóstico/prección/inferencia de los modelos

## 3. Aprendizaje Supervisado y Aprendizaje No-Supervisado

**Apredizaje Supervisado:** Modelos de inferencia/predicción/análisis derivados del conjunto de datos observado, donde las respuestás están fijadas en el comportamiento observado de los datos.

* Determinar si un paciente tendrá diabetes, dado el comportamiento de 1,000 pacientes observados y su desenlace sobre si tuvo o no tuvo diabetes (histórico clínico)
* Determinar la satisfacción de un cliente al recibir un servicio bancario, dada la satasfacción registrada en 10,000 clientes a lo largo de 2 años (histórico de encuestas de satisfacción)
* Determinar la probabilidad de supervivencia en el Titanic, dado el registro de 891 supervivientes y no supervivientes (histórico de los pasajeros del titanic y su supervivencia)

**Aprendizaje No-Supervisado:** Técnicas para construir nueva información derivada de conjuntos de datos no etiquetados, es decir, conjuntos de datos cuya respuesta no es explícita o parte del análisis.

* Determinar la similitud entre clientes que ven series en Netflix, dadas sus características
* Determinar el clúster de un producto dadas las características de los productos
* Representar 50 características en 2 componentes principales, reduciendo la información de 50 dimensiones a solo 2 dimensiones

## 4. Naturaleza Cuantitativa y Cualitativa

**Cuantitativa:** Una característica que representa información continua entre las observaciones, por ejemplo: pesos, factores, medidas, indicadores, métricas.

**Cualitativa:** Una característica que representa información discreta o relacional entre las observaciones, por ejemplo: pertenecia, efectos, capacidades, cualidades, relaciones.

**Numérica:** Una característica expresa en números (cualitativa 0/1 o cuantitativa -100...100)

**Categórica:** Una característica expresa en grupos/categorías/clases (cualitativas A/B/C/...)
