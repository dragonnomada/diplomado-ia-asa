```python
# Los comentarios sirven para documentar código dentro del código
# Se antepone un símbolo de # y lo siguiente es texto ignorado

# Varibale: memoria nombrada, o datos etiquetados
# - Tienen nombres específicos y sensibles a las minúsculas y mayúsculas
# - Aceptan el guión bajo _ y números, pero con restricciones
# - Existen concenciones o forma de nombrar las variables:
#   * Clases y objetos: Se suele usar la forma Húngura (ej. TipoAeronave)
#   * Variables comunes: Se suele usar la forma Serpiente (ej. resultado_esperado_operacion)
#   * Funciones y métodos: Se suele usar la forma Camello (ej. calcularIVACarritoCompras)
# - Hay restricciones:
#   * No pueden comenzar con números (ej. 123persona)
#   * No pueden contener símbolos salvo guión bajo (ej. $hola)

# Tipos de datos que se pueden almacenar: 
# - int (número entero): 123, 456, ...
# - float (número decimal): 123.456, 456.789,...
# - str (texto): "hola mundo", 'Anahí Pérez',...
# - bool (lógico): True (verdadero), False (Falso)
# - NoneType (nulo): None

nombre = "Lili María"
apellido_paterno = "Gómez"
apellido_materno = 'Aúnq'
edad = 23
peso = 65.2
casado = True
trabajo = False
puntaje_examen = None

# ! Define las variables sin efecto de salida
```

Formato de impresión -f-string o texto con formato

```python
#  Un texto puede incrustar resultados o valores dentro de un formato
# y transformase en un f-string anteponiendo la f

# {<expresión>} - Incrustación
print(f"Hola {nombre}")
```

```python
nombre_completo = f"{nombre}, {apellido_paterno} {apellido_materno}"

nombre_completo
```

```python
# Una incrustación se puede adaptar mediante :
# - :.<m>f determina para un decimal cuántos dígitos mostrar
# - :<n>.<m>f determina <n> caracteres, de ellos <m> decimales

print(f"""

Informe de la persona:
----------------------------
NOMBRE: {nombre}, {apellido_paterno} {apellido_materno}
EDAD: {edad} años
PESO: {peso:.2f} kg
CASADO: {["NO", "SI"][casado]}
TRABAJO: {["⛔️", "✅"][trabajo]}
-----------------------------------
PUNTAJE DE EXAMEN:     {"DESCONOCIDO" if puntaje_examen is None else puntaje_examen} pts.
""")
```

```python
# Condicional: SI -> ENTONCES

# BLOQUE DE INDENTACIÓN (Obligatorio)
# - Una forma de pertenencia a través de espacios o tabuladores
#   que indican si un línea pertenece al nivel superior

if 5 == 5: # BLOQUE :
    print("5 es igual a 5")
    print("Esto también se imprime")
```

```python
# Condicional: SI -> ENTONCES

# BLOQUE DE INDENTACIÓN (Obligatorio)
# - Una forma de pertenencia a través de espacios o tabuladores
#   que indican si un línea pertenece al nivel superior

if 5 == 5: # BLOQUE :
    print("5 es igual a 5")
    print("Esto también se imprime")
```

```python
if 5 > 3:
    if 5 < 10:
        print("5 está dentro de (3, 10)")
        print("otra línea")
```

```python
x = 6

if x > 3:
    if x < 10:
        print(f"{x} está dentro de (3, 10)")
```

```python
# input(<mensaje>) - Devuelve el texto capturado por el usuario
# int(•) - convierte • en entero (o falla)
x = int(input("Dame x:"))

if x > 3:
    if x < 10:
        print(f"{x} está dentro de (3, 10)")
```

```python
# Condicional doble (dual): SI -> ENTONCES | SINO -> ENTONCES

peso = float(input("Dame peso (kg):"))

if peso > 40:
    if peso < 100:
        print(f"El peso de {peso} kg es permitido")
    else:
        print(f"El peso de {peso} kg supera o iguala los 100kg")
        print("! No está permitido")
else:
    print(f"El peso de {peso} kg no logra superar los 40kg")
    print("! No está permitido")
```

```python
peso = float(input("Dame tu peso (kg):"))
altura = float(input("Dame tu estatura (m):"))

imc = peso / altura ** 2

print(f"El IMC es {imc:.2f}")

if imc < 24.5:
    print("BAJA DE PESO")
elif imc < 29.5:
    print("PESO NORMAL")
elif imc < 34.5:
    print("OBESIDAD TIPO I")
else:
    print("OBESIDAD TIPO II")
```
