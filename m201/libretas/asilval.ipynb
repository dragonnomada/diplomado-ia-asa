```python
peso = float(input("Dame tu peso (kg):"))
altura = float(input("Dame tu estatura (m):"))

imc = peso / altura ** 2

print(f"El IMC es {imc:.2f}")

if imc < 24.5:
    print("BAJA DE PESO")
elif imc < 29.5:
    print("PESO NORMAL")
elif imc < 34.5:
    print("OBESIDAD TIPO I")
else:
    print("OBESIDAD TIPO II")
```

```python
# ==========================================
# EJEMPLO: MONITOREO DE COMBUSTIBLE - ASA
# ==========================================

from datetime import datetime

# Datos simulados de estaciones
estaciones = [
    {"nombre": "MEX", "combustible": 15000, "limite": 12000},
    {"nombre": "CUN", "combustible": 8500, "limite": 10000},
    {"nombre": "GDL", "combustible": 5000, "limite": 6000},
    {"nombre": "MTY", "combustible": 18000, "limite": 15000},
    {"nombre": "PVR", "combustible": 3000, "limite": 5000},
]

print("=" * 55)
print("       SISTEMA DE MONITOREO DE COMBUSTIBLE")
print("       AEROPUERTOS Y SERVICIOS AUXILIARES")
print("=" * 55)

print(f"\nFecha y hora: {datetime.now():%d/%m/%Y %H:%M:%S}\n")

for estacion in estaciones:

    nombre = estacion["nombre"]
    nivel = estacion["combustible"]
    limite = estacion["limite"]

    porcentaje = (nivel / limite) * 100

    if porcentaje >= 100:
        estado = "ALERTA: NIVEL SUPERIOR AL LIMITE"
    elif porcentaje >= 70:
        estado = "NORMAL"
    else:
        estado = "PRECAUCION: NIVEL BAJO"

    print(f"Estación: {nombre}")
    print(f"Combustible disponible: {nivel:,} litros")
    print(f"Nivel de referencia: {limite:,} litros")
    print(f"Porcentaje: {porcentaje:.1f}%")
    print(f"Estado: {estado}")
    print("-" * 55)

print("\nMonitoreo finalizado.")
```

```python
# Función que determina si un punto está dentro de un circulo

# Circulo: radio, centro (ox, oy)
# Punto: px, py

def enCirculo(px, py, ox, oy, radio):
    dx = px - ox
    dy = py - oy
    
    d = (dx ** 2 + dy ** 2) ** 0.5
    
    return d <= radio

    enCirculo(1, 1, 0, 0, 2)
```

```python
def enCirculo(punto, circulo):
    x, y = punto
    ox, oy, radio = circulo
    
    dx = x - ox
    dy = y - oy
    
    d = (dx ** 2 + dy ** 2) ** 0.5
    
    return d <= radio
```

```python
enCirculo((1, 1), (0, 0, 2))
```

```python
punto = (1, 1)
circulo = (0, 0, 2)

enCirculo (punto, circulo)
```

```python
import random

def crearPersona(nombres, edadMinima, edadMaxima):
    nombre = random.choice(nombres)
    edad = random.randint(edadMinima, edadMaxima + 1)
    return {
        "nombre": nombre,
        "edad": edad
    }
```

```python
crearPersona(nombres, 18, 25)
```

```python
crearPersona(nombres, 18, 25)
```

```python
personas = []

for i in range(10):
    persona = crearPersona(nombres, 18, 100)
    personas.append(persona)

personas
```
