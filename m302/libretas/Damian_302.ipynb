## 2. Pandas - Series / Estadística / Visualización / Reportes (Continuo y Discreto)

```python
import pandas
```

```python
s1 = pandas.Series([23, 24, 28, 34, 23, 21, 40, 24], name="edades")

s1
```

```python
print(s1)
print(s1.values) # Arreglo de numpy
print(s1.min())
print(s1.max())
print(s1.argmin()) # índice del mínimo
print(s1.argmax()) # índice del máximo
print(s1.sum())
print(s1.mean()) # promedio / media
print(s1.var()) # varianza (desviación cuadrática)
print(s1.std()) # desviación (desviación lineal / raíz de la varianza)
print(s1.quantile(0.05)) # Inferior al 5% de población ordenada
print(s1.quantile(0.25)) # Q1 - Inferior al 25% de población ordenada
print(s1.quantile(0.50)) # Q2 - Mediano al 50% de población ordenada (mediana)
print(s1.quantile(0.75)) # Q3 - Superior al 75% de población ordenada
print(s1.quantile(0.95)) # Superior al 95% de población ordenada
print(s1.quantile(0.75) - s1.quantile(0.25)) # IQR - Rango intercuantílico
```

```python
s1 = pandas.Series([17.99, 18.13, 18.01, 19.25, 19.01]) # precios del dólar
s2 = pandas.Series([21.99, 22.03, 21.87, 22.80, 22.21]) # precios de la gasolina

print(s1)
print(s2)

print(s1.corr(s2)) # Correlación: 
                    # +1 -> Ambos crecen igual
                    # -1 -> Ambos crecen contrarios
                    # 0 -> No se explican los cambios entre ambos

# 0.8 -> Siguen una tendencia lineal positiva alta
# 0.5 -> Siguen una tendencia lineal positiva baja
# 0.2 -> Siguen una tendencia lineal positiva no significativa
# 0 -> No siguen ninguna tendencia lineal
# -0.2 -> Siguen una tendencia lineal inversa no significativa
# -0.5 -> Siguen una tendencia lineal inversa baja
# -0.8 -> Siguen una tendencia lineal inversa alta
```

```python
s1.plot.bar(alpha=0.5, color="blue") # Atrapados
s2.plot.bar(alpha=0.3, color="red") # Más altos
```

```python
s1.plot.line()
s2.plot.line()
```

```python
tipos_causas = ["A", "B", "C", "I", "U"]
mu_causas = [50, 800, 400, 100_000, 3_0000]
sigma_causas = [10, 120, 30, 5_000, 2_000]

causas = []
defunciones = []

print(list(zip(tipos_causas, mu_causas, sigma_causas)))

for tipo, mu, sigma in zip(tipos_causas, mu_causas, sigma_causas):
    print(tipo, mu, sigma)
    for i in range(1, 21):
        causa = f"{tipo}{i:03d}"
        causas.append(causa)
        n_casos = int(np.random.normal(mu, sigma)) # Limita a un rango
        if n_casos < 0:
            n_casos = 0
        defunciones.append(n_casos)

print(causas)
print(defunciones)

df1 = pandas.DataFrame([
    causas,
    defunciones,
], index=["Causa", "Defunciones"]).T

df1
```

```python
# DataFrame a partir de un list-frame

tipos_causas = ["A", "B", "C", "I", "U"]
mu_causas = [50, 800, 400, 100_000, 3_0000]
sigma_causas = [10, 120, 30, 5_000, 2_000]

registros = [] # Cada registro es listado

for tipo, mu, sigma in zip(tipos_causas, mu_causas, sigma_causas):
    for i in range(1, 21):
        causa = f"{tipo}{i:03d}"
        n_casos = int(np.random.normal(mu, sigma)) # Limita a un rango
        if n_casos < 0:
            n_casos = 0
        registros.append({
            "Causa": causa,
            "Defunciones": n_casos
        })

print(registros)

df2 = pandas.DataFrame(registros)

df2
```

> Concatena las mujeres de 1ra clase, seguidas de las mujeres de 3ra clase en las columnas

```python
titanic[["Sex", "Age", "Fare", "Survived"]]
```

```python
first_class_mask = titanic["Pclass"] == 1
third_class_mask = titanic["Pclass"] == 3
female_mask = titanic["Sex"] == "female"

pandas.concat([
    titanic[female_mask & first_class_mask][["Sex", "Age", "Fare", "Survived"]],
    titanic[female_mask & third_class_mask][["Sex", "Age", "Fare", "Survived"]]
])
```
