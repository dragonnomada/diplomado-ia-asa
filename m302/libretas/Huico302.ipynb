```python
# DataFrame a partir de matrices

tipos_causas = ["A", "B", "C", "I", "U"]
mu_causas = [50, 800, 400, 100_000, 3_0000]
sigma_causas = [10, 120, 30, 5_000, 2_000]

causas = [] # Columna 1
defunciones = [] # columna 2

print(list(zip(tipos_causas, mu_causas, sigma_causas)))

for tipo, mu, sigma in zip(tipos_causas, mu_causas, sigma_causas):
    print(tipo, mu, sigma)
    for i in range(1, 21):
        causa = f"{tipo}{i:03d}"
        causas.append(causa)
        n_casos = int(numpy.random.normal(mu, sigma)) # Limita a un rango
        if n_casos < 0:
            n_casos = 0
        defunciones.append(n_casos)

print(causas)
print(defunciones)

d6 = pandas.DataFrame([ # Matriz con las columnas de información (* Transpuesta)
    causas,
    defunciones,
], index=["Causa", "Defunciones"]).T

d6
```

```python
# DataFrame a partir de frame-list

d6b = pandas.DataFrame({
    "Causa": causas,
    "Defunciones": defunciones
})

d6b
```

```python
# DataFrame a partir de un list-frame

tipos_causas = ["A", "B", "C", "I", "U"]
mu_causas = [50, 800, 400, 100_000, 3_0000]
sigma_causas = [10, 120, 30, 5_000, 2_000]

registros = [] # Cada registro es listado

for tipo, mu, sigma in zip(tipos_causas, mu_causas, sigma_causas):
    for i in range(1, 21):
        causa = f"{tipo}{i:03d}"
        n_casos = int(numpy.random.normal(mu, sigma)) # Limita a un rango
        if n_casos < 0:
            n_casos = 0
        registros.append({
            "Causa": causa,
            "Defunciones": n_casos
        })

print(registros)

d7 = pandas.DataFrame(registros)

d7
```

Ejercicio E302.3 - Recuperar los que pagaron tarifas bajas¶
Aplicar una máscara para Fare que sea menor o igual a 8 libras
¿Cuántos son?
¿Cuál es la media/promedio de supervivencia en estos registros?
Pista: La supervivencia promedio para todos es titanic["Survived"].mean()

Pista2: La supervivencia promedio para la máscara es titanic[<máscara>]["Survived"].mean()

```python
titanic["Survived"].mean()
```

```python
# 1. Crear la máscara booleana para tarifas menores o iguales a 8 libras
mascara = titanic["Fare"] <= 8

# 2. Contar cuántos registros cumplen la condición
cantidad = mascara.sum()  # Alternativa: len(titanic[mascara])

# 3. Calcular el promedio de supervivencia en ese subconjunto
supervivencia_baja = titanic[mascara]["Survived"].mean()
supervivencia_total = titanic["Survived"].mean()

print(f"Pasajeros con tarifa <= 8: {cantidad}")
print(f"Supervivencia promedio (tarifa <= 8): {supervivencia_baja:.4f}")
print(f"Supervivencia promedio (general): {supervivencia_total:.4f}")
```

```python
titanic[titanic["Fare"] <= 8]["Survived"].mean()
```

```python
titanic[titanic["Fare"] > 8]["Survived"].mean()
```

```python
# Hipótesis:

# 1. La supervivencia en mujeres es distinta en promedio a la de hombres
print(titanic[titanic["Sex"] == "female"]["Survived"].mean())
print(titanic[titanic["Sex"] == "male"]["Survived"].mean())
print()

# 2. La supervivencia de los de 1ra clase es distinta en promedio a la de 3ra clase
print(titanic[titanic["Pclass"] == 1]["Survived"].mean())
print(titanic[titanic["Pclass"] == 3]["Survived"].mean())
print()

# 3. La supervivencia favorece a mujeres 1ra clase, y se contrasta al extremo de hombres 3ra clase
print(titanic[titanic["Sex"] == "female"][titanic["Pclass"] == 1]["Survived"].mean())
print(titanic[titanic["Sex"] == "female"][titanic["Pclass"] == 3]["Survived"].mean())
print()
print(titanic[titanic["Sex"] == "male"][titanic["Pclass"] == 1]["Survived"].mean())
print(titanic[titanic["Sex"] == "male"][titanic["Pclass"] == 3]["Survived"].mean())
```
