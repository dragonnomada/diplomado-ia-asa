```python
N = 100
M = 200

suma = 0
conteo = 0

for i in range(1, N + 1):
    for j in range(1,M + 1):
        if i % 3 == 0:
            if j % 5 == 0:
                suma = suma + (i + j)
                conteo = conteo + 1

#suma, conteo, suma / conteo
print(f"Suma: {suma} | Total: {conteo} | Promedio: {suma/conteo}")
```

```python
suma_e = 0
conteo_e = 0
diferencias = []

for edad in edades:
    suma_e += edad
    conteo_e += 1

promedio_e = suma_e / conteo_e

for edad in edades:
    diferencia = edad - promedio_e
    diferencias.append(diferencia)

print(f"Total: {conteo_e}")
print("-" * 70)
print(f"Suma: {suma_e}")
print(f"Promedio: {promedio_e}")
print("-" * 70)
print(f"Diferencias: {diferencias}")
```

```python
precios_originales = [22.89, 22.81, 21.99, 23.45, 22.17, 22.63, 21.88, 21.81]
precios_ordenados = []

n = len(precios_originales)

for i in range(n):
    precio_minimo = min(precios_originales)
    precios_ordenados.append(precio_minimo)
    precios_originales.remove(precio_minimo)

print(precios_originales)
print(precios_ordenados)
```
