```python
# Problema P301.3 - Búqueda y suma de los múltiplos de 3 y 5

# Suma i de 1 a N, Suma de j de 1 a M si i es múltiplo de 3 y j es múltiplo de 5
# (i=1, j=1) - No suma
# ...
# (i=3, j=1) - No suma
# ...
# (i=3, j=5) - Suma 3 + 5 = 8
# ...
# (i=6, j=5) - Suma 6 + 5 = 11
# ...
# (i=12, j=25) - Suma 12 + 25 = 37
# ...

N = 300 # i=1, 2, [3], 4, 5, [6], ..., 95, [96], 97, 98, [99], 300
M = 160 # j=1, 2, 3, 4, [5], 6, ..., [195], 196, 197, 198, 199, [300]

suma = 0
conteo = 0

for i in range(1, N + 1):
    for j in range(40, M + 1):
        # print(f"(i={i:3d}, j={j:3d})")
        if i % 7 == 0:
            if j % 8 == 0:
                # print(f"(i={i:3d}, j={j:3d}) -> i + j = {i + j}")
                suma = suma + (i + j)
                conteo = conteo + 1

# suma, conteo, suma / conteo
print(f"Suma: {suma} | Total: {conteo} | Promedio: {suma / conteo}")

edades =[34,56,45,26,28,46,34,76]
n= 0
suma = 0 
promedio = 0 
dif = 0 
for edad in edades:
    n = n+1
    suma = suma + edad

promedio = suma/n

diferencia = edades

for edad in diferencia:
    dif = edad - promedio
    diferencia.append(dif)

print (suma)
print (promedio)
```

```python
#Algoritmo de ordenamiento por eliminación del mínimo/máximo

price_org =[234, 344,123,865,34,456,10,6843]
price_ord=[]

n= len(price_org)

for i in range(n):
    price_min = min(price_org)
    price_ord.append(price_min)
    price_org.remove(price_min)

print(price_org)
print(price_ord)


```
