```python
# Problema P301.3 - Búsqueda y suma de los múltiplos de 3 y 5

'''
Suma i de 1 a N, suma de j de 1 a M si i es múltiplo de 3 y j es múltiplo de 5
(i=1, j=1) - no suma
...
(i=3, j=1) - no suma
...
(i=3, j=5) - suma 3+5=8
..
(i=6, j=5) - suma 6+5=11
...
(i=N, j=M)
'''
N = 100
M = 200

suma = 0
conteo = 0

for i in range(1, N+1):
    for j in range (1, M+1):
        #print(f"(i={i}, j={j})")
        if i % 3 == 0:
            if j % 5 == 0:
                suma += i + j
                conteo += 1

print(f"Suma: {suma} | Total: {conteo} | Promedio: {suma / conteo}")
```

```python
# Ejercicio P301.3 - Búsqueda y suma de los múltiplos de 7 y 8
# i de 1 hasta 300
# j de 40 hasta 160

N = 300
M = 160

suma = 0
conteo = 0

for i in range(1, N+1):
    for j in range (40, M+1):
        #print(f"(i={i}, j={j})")
        if i % 7 == 0:
            if j % 8 == 0:
                suma += i + j
                conteo += 1

print(f"Suma: {suma} | Total: {conteo} | Promedio: {suma / conteo}")
```

```python
# Ejercicio E301.4 - Determinar qué tanto se alejan las edades de su media

# 1. Sumar todas las edades
# 2. Calcular el promedio de las edades
# 3. Calcular las diferencias entre cada edad y el promedio

edades = [28, 32, 45, 31, 30, 45, 67, 32]
suma = 0
diferencias = []

for edad in edades:
    suma += edad

promedio = suma / len(edades)

for edad in edades:
    indicador = edad - promedio
    diferencias.append(indicador)

print(f"La suma de las edades es: {suma}") #1
print(f"El promedio de las edades es: {promedio}") #2
print(f"La diferencia entre cada edad y el promedio es: {diferencias}") #3
```
