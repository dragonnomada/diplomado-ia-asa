```python
# Problema P301.3 - Búqueda y suma de los múltiplos de 3 y 5

# Suma i de 1 a N, Suma de j de 1 a M si i es múltiplo de 3 y j es múltiplo de 5
# (i=1, j=1) - No suma
# ...
# (i=3, j=1) - No suma
# ...
# (i=3, j=5) - Suma 3 + 5 = 8
# ...
# (i=6, j=5) - Suma 6 + 5 = 11
# ...
# (i=12, j=25) - Suma 12 + 25 = 37
# ...

N = 100 # i=1, 2, [3], 4, 5, [6], ..., 95, [96], 97, 98, [99], 100
M = 200 # j=1, 2, 3, 4, [5], 6, ..., [195], 196, 197, 198, 199, [200]

suma = 0
conteo = 0

for i in range(1, N + 1):
    for j in range(1, M + 1):
        # print(f"(i={i:3d}, j={j:3d})")
        if i % 3 == 0:
            if j % 5 == 0:
                # print(f"(i={i:3d}, j={j:3d}) -> i + j = {i + j}")
                suma = suma + (i + j)
                conteo = conteo + 1

# suma, conteo, suma / conteo
print(f"Suma: {suma} | Total: {conteo} | Promedio: {suma / conteo}")
```

```python
# Ejercicio E301.3 - Búsqueda y suma de los múltiplos de 7 y 8

# Calcular la suma de las combinaciones de los múltiplos de 7 desde 1 hasta 300
# y los múltiplos de 8 desde 40 hasta 160
suma_combinaciones = 0
contador_combinaciones = 0

# Recorrer múltiplos de 7 desde 1 hasta 300
for i in range(1, 301):
    if i % 7 == 0:
        # Por cada múltiplo de 7, recorrer múltiplos de 8 desde 40 hasta 160
        for j in range(40, 161):
            if j % 8 == 0:
                suma_combinaciones += (i + j)
                contador_combinaciones += 1

print(f"Total de combinaciones: {contador_combinaciones}")
print(f"Suma de todas las combinaciones: {suma_combinaciones}")
```

```python
# Ejercicio E301.4 - Determinar qué tanto se alejan las edades de su media

# 1. Sumar todas las edades
suma_edades = 0
for edad in edades:
    suma_edades += edad

# 2. Calcular el promedio de las edades
promedio = suma_edades / len(edades)

print(f"Suma de edades: {suma_edades}")
print(f"Promedio (media): {promedio}\n")

# 3. Calcular las diferencias entre cada edad y el promedio
print("Diferencias respecto al promedio:")
for edad in edades:
    diferencia = edad - promedio
    print(f"Edad: {edad} -> Diferencia: {diferencia}")
```

```python
import matplotlib.pyplot as plt

# Lista de edades
etiquetas = [f"P{i+1}" for i in range(len(edades))]

# Cálculo del promedio
promedio = sum(edades) / len(edades)

# Crear la gráfica
plt.figure(figsize=(8, 5))
plt.axhline(promedio, color="red", linestyle="--", label=f"Promedio: {promedio:.1f}")

for i, edad in enumerate(edades):
    diferencia = edad - promedio
    # Línea vertical desde el promedio hasta la edad
    plt.vlines(i, promedio, edad, color="gray", linestyle="-", linewidth=2)
    # Punto de la edad
    plt.scatter(i, edad, color="blue", zorder=5)
    # Etiqueta con el valor y su diferencia
    plt.text(i, edad + 0.8, f"{edad} ({diferencia:+.1f})", ha="center")

plt.xticks(range(len(edades)), etiquetas)
plt.ylabel("Edad")
plt.title("Alejamiento de las edades respecto a su media")
plt.legend()
plt.grid(axis="y", alpha=0.3)
plt.show()
```

```python
# Ejercicio E301.4 - Determinar qué tanto se alejan las edades de su media

# 1. Sumar todas edades
suma = 0 # sum(edades)
total = 0 # len(edades)
for edad in edades:
    suma += edad
    total += 1
# 2. Calcular el promedio de las edades
promedio = suma / total # suma / len(edades)
# 3. Calcular las diferencias entre cada edad y el promedio
diferencias = []
for edad in edades: # ! Cuidado no pueden recorrer una secuencia y adentro agregar más elementos
    diferencia = edad - promedio
    diferencias.append(diferencia)

print(f"Total:       {total}")
print("-" * 60)
print(f"Suma:        {suma}")
print(f"Promedio:    {promedio}")
print("-" * 60)
print(f"Diferencias: {diferencias}")
```

```python
# Algoritmo de Ordenamiento por eliminación del mínimo/máximo

precios_originales = [22.89, 22.81, 21.99, 23.45, 22.17, 22.63, 21.88, 21.81]
precios_ordenados = []

# Guardamos la cantidad total de elementos (n)
n = len(precios_originales)

# 4. Repetir por cada precio (n-veces) la misma operación
for _ in range(n):
    # 1. Buscar el precio mínimo
    minimo = min(precios_originales)
    
    # 2. Copiar ese precio mínimo en los precios ordenados
    precios_ordenados.append(minimo)
    
    # 3. Quitar el precio mínimo de los precios originales
    precios_originales.remove(minimo)

print("Precios ordenados:", precios_ordenados)
```

```python
precios = [22.89, 22.81, 21.99, 23.45, 22.17, 22.63, 21.88, 21.81, 23.45]

print(precios)
print("=" * 80)

n = len(precios)

# 1. Dividir precios de la mitad izquierda y la mitad derecha
p = int(n / 2)
precios_izquierda = precios[0:p] # [:p]
precios_derecha = precios[p:n] # [p:]

print("1. Dividir los precios en 2")
print()
print(precios_izquierda)
print(precios_derecha)
print()
print("-" * 80)

# 2. Reportar los 3 precios más bajos y los 3 más altos
precios_copia = precios[:] # Recupera los precios desde el inicio hasta el final
precios_ordenados = sorted(precios_copia)

print("2. Reportar el top/primeros precios bajos y altos")
print()
print(precios_copia)
print(precios_ordenados)
print()
print(precios_ordenados[:3])
print(precios_ordenados[-3:])
print()
print("-" * 80)

# 3. Precios pares e impares

precios_pares = precios[::2] # [0:n:2] - iterador par
precios_impares = precios[1::2] # [1:n:2] - iterador impar

print("3. Precios pares e impares")
print()
print(precios_pares)
print(precios_impares)
print()
print(sum(precios_pares) / len(precios_pares))
print(sum(precios_impares) / len(precios_impares))
print(sum(precios) / len(precios))
print()
print("-" * 80)
```

```python
# Problema de agotamiento de existencias

productos = [
    { "producto_id": 1, "nombre": "Coca Cola", "precio": 28.5, "existencias": 240 },
    { "producto_id": 2, "nombre": "Gansito", "precio": 24.8, "existencias": 32 },
    { "producto_id": 3, "nombre": "Galletas Marías", "precio": 17.5, "existencias": 23 },
    { "producto_id": 4, "nombre": "Chocotorros", "precio": 25.5, "existencias": 0 },
    { "producto_id": 5, "nombre": "Halls", "precio": 16.45, "existencias": 3 },
]

venta_producto = [
    { "venta_id": 1001, "producto_id": 1, "cantidad": 3 },
    { "venta_id": 1001, "producto_id": 1, "cantidad": 1 },
    { "venta_id": 1001, "producto_id": 2, "cantidad": 1 },
    { "venta_id": 1002, "producto_id": 3, "cantidad": 5 },
    { "venta_id": 1002, "producto_id": 4, "cantidad": 4 },
    { "venta_id": 1002, "producto_id": 5, "cantidad": 2 },
    { "venta_id": 1002, "producto_id": 1, "cantidad": 1 },
]

# 1. Determinar para cada producto cuántas cantidades se llevaron en cada venta
productos_ids = []
for producto in productos:
    if not producto["producto_id"] in productos_ids:
        productos_ids.append(producto["producto_id"])
productos_ids = sorted(productos_ids)

ventas_ids = []
for venta in venta_producto:
    if not venta["venta_id"] in ventas_ids:
        ventas_ids.append(venta["venta_id"])
ventas_ids = sorted(ventas_ids)

producto_venta_cantidades = {}

for producto_id in productos_ids:
    for venta_id in ventas_ids:
        for venta in venta_producto:
            if venta["venta_id"] == venta_id and venta["producto_id"] == producto_id:
                if not producto_id in producto_venta_cantidades:
                    producto_venta_cantidades[producto_id] = {}
                if not venta_id in producto_venta_cantidades[producto_id]:
                    producto_venta_cantidades[producto_id][venta_id] = 0
                producto_venta_cantidades[producto_id][venta_id] += venta["cantidad"]

print(producto_venta_cantidades)

producto_totales = {}
```
