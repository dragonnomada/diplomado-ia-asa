```python
# Problema P301.2 - Hacer sonar diferentes alertas

# Una alarma es sensible a la temperatura y la humedad
# La alarma debe indicar estados críticos, estados normales y estos bajos
# ^ R1. Cuando la temperatura es crítica y la humedad es crítica debe sonar la alarma de evacuación
# ^ R2. Si la temperatura y humedad son bajas debe sonar la alarma de alerta
# ^ R3. Si la temperatura y humedad son normales debe sonar la alarma de normalidad
# * R4. Cuando la temperatura es crítica debe sonar la alarma de incendio
# * R5. Cuando la humedad es crítica debe sonar la alarma de inundación
# * R6. Si la temperatura es baja debe sonar la alarma de frío
# * R7. Si la humedad es baja debe sonar la alarma de sequía

# Variables
temperatura = float(input("Temperatura: ")) # Lectura y conversión a decimal desde el usuario
humedad = float(input("Humedad: ")) # Lectura y conversión a decimal desde el usuario

umbral_temperatura_alta = 45
umbral_temperatura_baja = 0
umbral_humedad_alta = 80
umbral_humedad_baja = 20

r1 = None
r2 = None
r3 = None
r4 = None
r5 = None
r6 = None
r7 = None

es_temperatura_alta = None
es_temperatura_baja = None
es_temperatura_normal = None
es_humedad_alta = None
es_humedad_baja = None
es_humedad_normal = None

# Operaciones/Cálculos intermedios
es_temperatura_alta = temperatura >= umbral_temperatura_alta
es_temperatura_baja = temperatura <= umbral_temperatura_baja
es_temperatura_normal = (not es_temperatura_alta) and (not es_temperatura_baja)

es_humedad_alta = humedad >= umbral_humedad_alta
es_humedad_baja = humedad <= umbral_humedad_baja
es_humedad_normal = (not es_humedad_alta) and (not es_humedad_baja)

r1 = es_temperatura_alta and es_humedad_alta
r2 = es_temperatura_baja and es_humedad_baja
r3 = es_temperatura_normal and es_humedad_normal
r4 = es_temperatura_alta
r5 = es_humedad_alta
r6 = es_temperatura_baja
r7 = es_humedad_baja

# Resultados y reportes
if r1:
    print("ALARMA DE EVACUACIÓN")
else:
    if r2:
        print("ALARMA DE ALERTA")
    else:
        if r3:
            print("ALARMA DE NORMALIDAD")
        else:
            if r4:
                print("ALERTA DE INCENDIO")
            else:
                if r5:
                    print("ALERTA DE INUNDACIÓN")
                else:
                    if r6:
                        print("ALERTA DE FRÍO")
                    else:
                        if r7:
                            print("ALERTA DE SEQUÍA")
                        else:
                            print("ERROR LÓGICO")
```

```python
# Problema P301.3 - Búqueda y suma de los múltiplos de 3 y 5

# Suma i de 1 a N, Suma de j de 1 a M si i es múltiplo de 3 y j es múltiplo de 5
# (i=1, j=1) - No suma
# ...
# (i=3, j=1) - No suma
# ...
# (i=3, j=5) - Suma 3 + 5 = 8
# ...
# (i=6, j=5) - Suma 6 + 5 = 11
# ...
# (i=12, j=25) - Suma 12 + 25 = 37
# ...

N = 100 # i=1, 2, [3], 4, 5, [6], ..., 95, [96], 97, 98, [99], 100
M = 200 # j=1, 2, 3, 4, [5], 6, ..., [195], 196, 197, 198, 199, [200]

suma = 0
conteo = 0

for i in range(1, N + 1):
    for j in range(1, M + 1):
        # print(f"(i={i:3d}, j={j:3d})")
        if i % 3 == 0:
            if j % 5 == 0:
                # print(f"(i={i:3d}, j={j:3d}) -> i + j = {i + j}")
                suma = suma + (i + j)
                conteo = conteo + 1

# suma, conteo, suma / conteo
print(f"Suma: {suma} | Total: {conteo} | Promedio: {suma / conteo}")
```

```python
# Ejercicio E301.4 - Determinar qué tanto se alejan las edades de su media

# 1. Sumar todas edades
suma = 0 # sum(edades)
total = 0 # len(edades)
for edad in edades:
    suma += edad
    total += 1
# 2. Calcular el promedio de las edades
promedio = suma / total # suma / len(edades)
# 3. Calcular las diferencias entre cada edad y el promedio
diferencias = []
for edad in edades: # ! Cuidado no pueden recorrer una secuencia y adentro agregar más elementos
    diferencia = edad - promedio
    diferencias.append(diferencia)

print(f"Total:       {total}")
print("-" * 60)
print(f"Suma:        {suma}")
print(f"Promedio:    {promedio}")
print("-" * 60)
print(f"Diferencias: {diferencias}")
```

```python
# Algoritmo de Ordenamiento por eliminación del mínimo/máximo

# ITERACIÓN 1:
# ANTES...
# ORIGINALES: 22.89, 22.81, 21.99, 23.45, 22.17, 22.63, 21.88, 21.81
# ORDENADOS: 
# DESPUÉS...
# ORIGINALES: 22.89, 21.99, 23.45, 22.17, 22.63, 21.88, 21.81
# ORDENADOS: 22.81
# ITERACIÓN 2:
# ANTES...
# ORIGINALES: 22.89, 21.99, 23.45, 22.17, 22.63, 21.88, 21.81
# ORDENADOS: 22.81
# DESPUÉS...
# ORIGINALES: 22.89, 21.99, 23.45, 22.17, 22.63, 21.88
# ORDENADOS: 22.81, 21.81
# ITERACIÓN 3:
# ANTES...
# ORIGINALES: 22.89, 21.99, 23.45, 22.17, 22.63, 21.88
# ORDENADOS: 22.81, 21.81
# DESPUÉS...
# ORIGINALES: 22.89, 21.99, 23.45, 22.17, 22.63
# ORDENADOS: 22.81, 21.81, 21.88
# ...
# ORIGINALES: 22.89, 23.45
# ORDENADOS: 22.81, 21.81, 21.88, 21.99, 22.17, 22.63
# ORIGINALES: 23.45
# ORDENADOS: 22.81, 21.81, 21.88, 21.99, 22.17, 22.63, 22.89
# ORIGINALES: 
# ORDENADOS: 22.81, 21.81, 21.88, 21.99, 22.17, 22.63, 22.89, 23.45

# 1. Buscar el precio mínimo
# 2. Copiar ese precio mínimo en los precios ordenados
# 3. Quitar el precio mínimo de los precios originales 
#    (por que sino siempre nos devolvería el mismo mínimo)
# 4. Repetir por cada precio (n-veces) la misma operación

precios_originales = [22.89, 22.81, 21.99, 23.45, 22.17, 22.63, 21.88, 21.81]
precios_ordenados = []

n = len(precios_originales)

# Repetir n-veces
for i in range(n):
    # 1. Recuperar el mínimo (orginales)
    precio_minimo = min(precios_originales) # ?
    # 2. Copiamos el mínimo (ordenados)
    precios_ordenados.append(precio_minimo) # ?
    # 2. Eliminamos el mínimo (originales)
    precios_originales.remove(precio_minimo) # ?

print(precios_originales)
print(precios_ordenados)
```

```python
precios = [22.89, 22.81, 21.99, 23.45, 22.17, 22.63, 21.88, 21.81, 23.45]

print(precios)
print("=" * 80)

n = len(precios)

# 1. Dividir precios de la mitad izquierda y la mitad derecha
p = int(n / 2)
precios_izquierda = precios[0:p] # [:p]
precios_derecha = precios[p:n] # [p:]

print("1. Dividir los precios en 2")
print()
print(precios_izquierda)
print(precios_derecha)
print()
print("-" * 80)

# 2. Reportar los 3 precios más bajos y los 3 más altos
precios_copia = precios[:] # Recupera los precios desde el inicio hasta el final
precios_ordenados = sorted(precios_copia)

print("2. Reportar el top/primeros precios bajos y altos")
print()
print(precios_copia)
print(precios_ordenados)
print()
print(precios_ordenados[:3])
print(precios_ordenados[-3:])
print()
print("-" * 80)

# 3. Precios pares e impares

precios_pares = precios[::2] # [0:n:2] - iterador par
precios_impares = precios[1::2] # [1:n:2] - iterador impar

print("3. Precios pares e impares")
print()
print(precios_pares)
print(precios_impares)
print()
print(sum(precios_pares) / len(precios_pares))
print(sum(precios_impares) / len(precios_impares))
print(sum(precios) / len(precios))
print()
print("-" * 80)
```

```python
# Problema de agotamiento de existencias

productos = [
    { "producto_id": 1, "nombre": "Coca Cola", "precio": 28.5, "existencias": 240 },
    { "producto_id": 2, "nombre": "Gansito", "precio": 24.8, "existencias": 32 },
    { "producto_id": 3, "nombre": "Galletas Marías", "precio": 17.5, "existencias": 23 },
    { "producto_id": 4, "nombre": "Chocotorros", "precio": 25.5, "existencias": 0 },
    { "producto_id": 5, "nombre": "Halls", "precio": 16.45, "existencias": 3 },
]

venta_producto = [
    { "venta_id": 1001, "producto_id": 1, "cantidad": 3 },
    { "venta_id": 1001, "producto_id": 1, "cantidad": 1 },
    { "venta_id": 1001, "producto_id": 2, "cantidad": 1 },
    { "venta_id": 1002, "producto_id": 3, "cantidad": 5 },
    { "venta_id": 1002, "producto_id": 4, "cantidad": 4 },
    { "venta_id": 1002, "producto_id": 5, "cantidad": 2 },
    { "venta_id": 1002, "producto_id": 1, "cantidad": 1 },
]

# 1. Determinar para cada producto cuántas cantidades se llevaron en cada venta
productos_ids = []
for producto in productos:
    if not producto["producto_id"] in productos_ids:
        productos_ids.append(producto["producto_id"])
productos_ids = sorted(productos_ids)

ventas_ids = []
for venta in venta_producto:
    if not venta["venta_id"] in ventas_ids:
        ventas_ids.append(venta["venta_id"])
ventas_ids = sorted(ventas_ids)

producto_venta_cantidades = {}

for producto_id in productos_ids:
    for venta_id in ventas_ids:
        for venta in venta_producto:
            if venta["venta_id"] == venta_id and venta["producto_id"] == producto_id:
                if not producto_id in producto_venta_cantidades:
                    producto_venta_cantidades[producto_id] = {}
                if not venta_id in producto_venta_cantidades[producto_id]:
                    producto_venta_cantidades[producto_id][venta_id] = 0
                producto_venta_cantidades[producto_id][venta_id] += venta["cantidad"]

print(producto_venta_cantidades)

producto_totales = {}
producto_total_ventas = {}

for producto_id in producto_venta_cantidades:
    if not producto_id in producto_total_ventas:
        producto_total_ventas[producto_id] = 0
    for venta_id in producto_venta_cantidades[producto_id]:
        producto_total_ventas[producto_id] += 1
        if not producto_id in producto_totales:
            producto_totales[producto_id] = 0
        producto_totales[producto_id] += producto_venta_cantidades[producto_id][venta_id]

print(producto_totales)
print(producto_total_ventas)

producto_promedio = {}

for producto_id in productos_ids:
    total_vendidos = producto_totales[producto_id]
    total_ventas = producto_total_ventas[producto_id]
    promedio = total_vendidos / total_ventas
    producto_promedio[producto_id] = promedio

print(producto_promedio)
```

```python
ventas_promedio_al_dia = 23

for producto_id in producto_promedio:
    venta_promedio = producto_promedio[producto_id]
    ventas_esperadas = venta_promedio * ventas_promedio_al_dia
    print(f"#{producto_id:04d} - Venta promedio: {venta_promedio} | Venta diaria*: {ventas_esperadas}")

    indicador = None

    for producto in productos:
        if producto["producto_id"] == producto_id:
            existencias = producto["existencias"]
            if ventas_esperadas == 0:
                indicador = 0
            else:
                indicador = existencias / ventas_esperadas
            print(f"        --------------------------------------------")
            print(f"        Existencias: {existencias} | Días de existencia: {indicador:.1f}")

    print("=" * 80)
```
